## Financial Dashboard

#### Company Profitability & Cost Analysis

- KPI cards: Revenue, Cost, Profit, Profit Margin
- Interactive line chart: Revenue, Cost, Profit trends
- Interactive bar charts: Profit Margin & Cost-to-Revenue Ratio

    

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px

# -----Sample Financial Dataset----------------------------
data = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Revenue": [120000, 135000, 150000, 160000],
    "Cost": [80000, 90000, 95000, 100000]
}
df = pd.DataFrame(data)

# Calculate Profit and Ratios
df["Profit"] = df["Revenue"] - df["Cost"]
df["Profit_Margin"] = df["Profit"] / df["Revenue"]
df["Cost_to_Revenue"] = df["Cost"] / df["Revenue"]

# -------------------- Summary KPIs ----------------------------
kpi_metrics = {
    "Total Revenue": df["Revenue"].sum(),
    "Total Cost": df["Cost"].sum(),
    "Total Profit": df["Profit"].sum(),
    "Average Profit Margin": df["Profit_Margin"].mean()
}

# KPI Cards (using Indicator traces)
fig_kpi = make_subplots(
    rows=1, cols=4,
    specs=[[{"type": "indicator"}]*4]
)

fig_kpi.add_trace(go.Indicator(
    mode="number",
    value=kpi_metrics["Total Revenue"],
    title={"text": "Total Revenue"},
    number={"prefix": "$"}
), row=1, col=1)

fig_kpi.add_trace(go.Indicator(
    mode="number",
    value=kpi_metrics["Total Cost"],
    title={"text": "Total Cost"},
    number={"prefix": "$"}
), row=1, col=2)

fig_kpi.add_trace(go.Indicator(
    mode="number",
    value=kpi_metrics["Total Profit"],
    title={"text": "Total Profit"},
    number={"prefix": "$"}
), row=1, col=3)

fig_kpi.add_trace(go.Indicator(
    mode="number",
    value=kpi_metrics["Average Profit Margin"],
    title={"text": "Avg Profit Margin"},
    number={"suffix": "%", "valueformat": ".0%"}
), row=1, col=4)

fig_kpi.update_layout(height=250, template="plotly_white", title="Executive Financial KPIs")
fig_kpi.show()
```

```python
# --------- Line Chart: Revenue, Cost, Profit ----------------------------
fig_line = go.Figure()
fig_line.add_trace(go.Scatter(x=df["Quarter"], y=df["Revenue"], mode="lines+markers", name="Revenue", line=dict(color="green", width=3)))
fig_line.add_trace(go.Scatter(x=df["Quarter"], y=df["Cost"], mode="lines+markers", name="Cost", line=dict(color="red", width=3)))
fig_line.add_trace(go.Scatter(x=df["Quarter"], y=df["Profit"], mode="lines+markers", name="Profit", line=dict(color="blue", width=3)))

fig_line.update_layout(
    title="Quarterly Revenue, Cost, and Profit",
    xaxis_title="Quarter",
    yaxis_title="Amount ($)",
    template="plotly_white",
    hovermode="x unified"
)
fig_line.show()
```

```python
# ------------------------ Bar Charts for  Profit Margin & Cost-to-Revenue ----------------------------
# Profit Margin
fig_pm = px.bar(
    df, x="Quarter", y="Profit_Margin",
    text=df["Profit_Margin"].apply(lambda x: f"{x:.0%}"),
    color="Profit_Margin",
    color_continuous_scale="Viridis",
    title="Quarterly Profit Margin"
)
fig_pm.update_layout(yaxis_tickformat=".0%")
fig_pm.show()

# Cost-to-Revenue Ratio
fig_cr = px.bar(
    df, x="Quarter", y="Cost_to_Revenue",
    text=df["Cost_to_Revenue"].apply(lambda x: f"{x:.0%}"),
    color="Cost_to_Revenue",
    color_continuous_scale="Magma",
    title="Quarterly Cost-to-Revenue Ratio"
)
fig_cr.update_layout(yaxis_tickformat=".0%")
fig_cr.show()
```
